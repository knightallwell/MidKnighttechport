```python

tickers = {
    "Apple": "AAPL",
    "Nvidia": "NVDA",
    "Microsoft": "MSFT",
    "Amazon": "AMZN",
    "Meta": "META",
    "Google": "GOOGL",
    "GooG": "GOOG",
    "Tesla": "TSLA",
    "Berkshire": "BRK-B",
    "UnitedHealth": "UNH"
}

data = {}
for company, ticker in tickers.items():
  data[company] = yf.Ticker(ticker).history(period="max")

```

```python
tickers = ' '.join(tickers.keys())

dfprimesp500 = yf.download(tickers, period="max")
```

```python

df = pd.concat(data, axis=1, keys=data.keys())

df.columns = ['_'.join(col).strip() for col in df.columns.values]
```
